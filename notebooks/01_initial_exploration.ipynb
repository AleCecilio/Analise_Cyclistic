## 1. Importação das Bibliotecas

```python
# Imports

import pandas as pd
```

---

## 2. Carregamento dos Dados

```python
# Cria string com data de identificação do arquivo/variavel

def str_date(i):
    if i < 4:
        year = 2025
        month = i + 9
    else:
        year = 2026
        month = i - 3

    return f"{year}{month:02d}"
```

```python
# Cria Dicionario de com os 12 Datasets 

datasets = {}

for i in range(12):
    
    file_name = f"{str_date(i)}-divvy-tripdata.csv"

    datasets[f"{str_date(i)}"] = pd.read_csv(
        f"../data/raw/{file_name}"
    )
```

---

## 3. Exploração Inicial dos Datasets

```python
for name, df in datasets.items():
    
    print(f"\n{'=' * 50}")
    print(f"DATASET: {name}")
    print(f"{'=' * 50}")

    display(df.head())
    print('\n')
```

---

## 4. Estrutura e Variáveis

```python
# Relatório do Shape de cada DataFrame 

for name, df in datasets.items():

    print(f"\n{'=' * 50}")
    print(f"DATASET: {name}")
    print(f"{'=' * 50}")
    
    print(f"Linhas: {df.shape[0]:,}")
    print(f"Colunas: {df.shape[1]}")
```

```python
# Verificando tipos das Variaveis 

columns_type = {}

for name, df in datasets.items():

    columns = df.columns
    
    print(f"\n{'=' * 50}")
    print(f"DATASET: {name}")
    print(f"{'=' * 50}")

    for i in range(12): 

        columns_type[name] = {columns[i] : df[columns[i]].dtype}

        print(f"{columns[i]} : {df[columns[i]].dtype}")
```

---

## 5. Comparação entre os Arquivos Mensais

```python
# Checando se todos os DataFrames possuem as mesmas colunas 
# Se 'False' todas as colunas são iguais
 
columns = {}

for name, df in datasets.items():
    columns[name] = list(df.columns)

different_columns = len(set(map(tuple, columns.values()))) > 1

print(different_columns)
```

```python
# Checando se alguma incongluencia nos tipos de variaveis em cada DataFrame
# Se 'False' todos os tipos são iguais para todas as colunas de todos dfs

different_columns_type = len(set(map(tuple, columns_type.values()))) > 1
print(different_columns_type)
```

---

## 6. Concatenação dos Datasets

```python
# Concatenado DataFrames

df = pd.concat(datasets.values(), ignore_index=True)
```

```python
# Checando a nova estrutura do Dataframe com todos os meses

df.shape
```

```python
# Explorando DataFrame com todos os meses

df.head()
```

```python
# Comparando DataFrame com os 12 Meses em relação aos DataFrames De Meses Individuais

individual_total = sum(len(dataset) for dataset in datasets.values())

print(f"Total dos datasets: {individual_total:,}")
print(f"Total após concatenação: {len(df):,}")
print(f"Os valores são iguais? {individual_total == len(df)}")
```

## 7. Salvando os Dados Consolidados

```python
# Salvando .csv na pasta processed

df.to_csv("../data/processed/cyclistic_202509_202608.csv", index=False)
```
